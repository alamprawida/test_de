```python
def generate_sn (row) :
    no = row['Number']
    let = row['Letter']

    # bagian 1
    if no % 2 == 0 :
        genap = no // 2
        bag_1 = let * genap
    else : 
        bag_1 = let

    # bagian 2
    bag_2 = 2 ** no

    # bagian 3
    bag_3 = f"{no}{let}"

    # bagian 4 (using modulo : dibagi habis, kalo ngk bisa dibagi ngk usah dibagi)
    bag_4 = no % 3

    # bagian 5
    if no % 3 == 1 : 
        bag_5 = "R"
    elif no % 3 == 2 : 
        bag_5 = "RR"
    elif no % 3 == 0 : 
        bag_5 = "O"
    else :
        bag_5 = "Error"

    return f"{bag_1} - {bag_2} - {bag_3} - {bag_4} - {bag_5}"
```

```python
df_result = raw.copy()
df_result['Serial Number'] = df_result.apply(generate_sn, axis=1)
df_result
```
